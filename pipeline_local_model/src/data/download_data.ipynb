# RewardBench

```python
preference_name = "rewardbench"
ds = list(load_dataset("allenai/reward-bench", split="filtered", cache_dir="/home/zhangqiyuan/qiyuan/tts_for_grm/data/input"))
random.shuffle(ds)
customized_ds = []
for case in ds:
    if random.uniform(0,1) < 0.5:
        label = 0
        response_a = case["chosen"]
        response_b = case["rejected"]
    else:
        label = 1
        response_a = case["rejected"]
        response_b = case["chosen"]
    case_template = {
        "instruction": case["prompt"],
        "response_a": response_a,
        "response_b": response_b,
        "label": label
    }
    customized_ds.append(case_template)
print(len(customized_ds))
write_output(os.path.join(SAMPLE_ROOT, preference_name+".json"), customized_ds)
```

```python
preference_name = "judgebench"
ds = list(load_dataset("ScalerLab/JudgeBench", split="gpt", cache_dir="/home/zhangqiyuan/qiyuan/tts_for_grm/data/input"))
random.shuffle(ds)
customized_ds = []
for case in ds:

    label = 0 if ">" in case["label"] else 1
    response_a = case["response_A"]
    response_b = case["response_B"]
    
    case_template = {
        "instruction": case["prompt"],
        "response_a": response_a,
        "response_b": response_b,
        "label": label
    }
    customized_ds.append(case_template)
print(len(customized_ds))
write_output(os.path.join(SAMPLE_ROOT, preference_name+".json"), customized_ds)
```
